```python
import os
from dotenv import load_dotenv

from scraper import crawl_website,fetch_page

from IPython.display import Markdown, display
from openai import OpenAI
import requests
```

```python
openai = OpenAI(
    base_url="http://localhost:11434/v1",
    api_key="ollama"
)
```

```python
MODEL = "llama3.1"
```

```python
system_prompt = """
You are a helpful assistant that analyzes website content.

Provide a short, clear summary of the website.
Ignore navigation menus, buttons, and irrelevant content.

Respond in markdown.
"""

user_prompt_prefix = """
Here are the contents of a website.

Please provide a useful summary of this website.

Website contents:

"""
```

```python
def messages_for(website):
    return [
        {
            "role": "system",
            "content": system_prompt
        },
        {
            "role": "user",
            "content": user_prompt_prefix + website
        }
    ]
```

```python
def summarize(url):
    website = fetch_page(url)

    response = openai.chat.completions.create(
        model=MODEL,
        messages=messages_for(website)
    )

    return response.choices[0].message.content
```

```python
def display_summary(url):
    summary = summarize(url)
    display(Markdown(summary))
```

```python
system_prompt = """
You are a helpful assistant that analyzes information from the
Texas Tech University Health Sciences Center El Paso website.

Answer questions using ONLY the website information provided to you.

If the information is not present in the provided website content,
say that you could not find the information.

Be clear and concise.

Respond in markdown.
"""
```

```python
user_prompt_prefix = """
Here is information from the TTUHSC El Paso website.

Use this information to answer the user's question.

Website information:

"""
```

```python
def messages_for(website_text):
    return [
        {
            "role": "system",
            "content": system_prompt
        },
        {
            "role": "user",
            "content": user_prompt_prefix + website_text
        }
    ]
```

```python
BASE_URL = 'url'
```

```python
def summarize(url):

    page = fetch_page(url)

    if page is None:
        return "Could not fetch the website."

    website_text = (
        "Title: " + page["title"] + "\n\n"
        "URL: " + page["url"] + "\n\n"
        "Content:\n" + page["text"]
    )

    response = openai.chat.completions.create(
        model=MODEL,
        messages=messages_for(website_text)
    )

    return response.choices[0].message.content
```

```python
from IPython.display import Markdown, display

def display_summary(url):
    summary = summarize(url)
    display(Markdown(summary))
```

```python
display_summary("https://ttuhscep.edu/")
```

```python
question = input("Ask a question about TTUHSC El Paso: ")

page = fetch_page("https://ttuhscep.edu/")

website_text = (
    "Title: " + page["title"] + "\n\n"
    "URL: " + page["url"] + "\n\n"
    "Website content:\n" + page["text"]
)

messages = [
    {
        "role": "system",
        "content": """
You are a helpful assistant for Texas Tech Health El Paso.

Answer the user's question using ONLY the website information
provided below.

If the answer cannot be found in the provided information,
say that you could not find it.

Do not invent information.
"""
    },
    {
        "role": "user",
        "content": f"""
Website information:

{website_text}

User question:

{question}
"""
    }
]

response = openai.chat.completions.create(
    model=MODEL,
    messages=messages
)

print(response.choices[0].message.content)
```
